### Huffman coding


1. __Initialization__: put all symbols on the list sorted according to their frequency counts.
2. __Repeat__ until the list has only one symbol left.
    * From the list, pick two symbols with the lowest frequency counts. From a Huffman subtree that has these two symbols as child nodes and create a parent node for them.
    * Assign the sum of the children’s frequency counts to the parent and insert it into the list, such that the order is maintained.
    * Delete the children from the list.
3. __Assign a codeword__ for each leaf based on the path from the root.

The above pseudocode can be __summarized__ as:
1. __Sort symbols__ by their probabilities.
2. __Pick__ two symbols with the __lowest probabilities__.
3. __Merge__ the two symbols and __remove__ the chosen symbols.
When we have only one symbol left. __Traverse__ the built binary tree and __assign '0' or '1'__.

There are many ways to implement Huffman coding. Among them, using a tree structure is one of the most common way. However, using a tree is not that intuitive to see how the algorithm works.
There are some options to implement Huffman code.
1. Using `tree` structure (Need to implement tree first)
2. Using `heapq` structure in Python
3. Using `recursion` (Since 1 to 3 steps in the summary and bit assigning performs repeatedly)
4. Using `two dictionaries` (One for symbol and prob., another for symbol and code mapping)

The 4th one can be the most intuitive way to implement Huffman code and very similar procedure as we saw in the lecture slide.

This is an example of how to use dictionaries for Huffman coding.

#### (1) Write a Huffman code generating function HuffmanCode() according to the following conditions. And print the dictionary of symbol and code.

#### Condition:
* Input arg: symbol and probability dictionary
* Output (return value): symbol and code dictionary
* When sorting symbols with the same probabilities, Do <font color=red>__NOT__</font> change the order. (e.g., {A:0.1, B:0.1}).
* Do <font color=red>__NOT__</font> use `tree`, `heapq` structure, or `recursion`.
* The order used in the lecture slides is in __ascending order from the bottom__, we can interpret this as a transposed vecter of in __ascending order from the left__.
* When you assign bits = {0, 1} to symbol, apply the same rule as the lecture slides, i.e., `1` to `top` and `0` to `bottom`.  
* Don't need to repeat until **one symbol left** (this is tree based algorithm). In our case, **repeat until two symbols left**
* Input dictionary: <br> 
    1. {'A': 0.1, 'B': 0.19, 'C': 0.21, 'D': 0.3, 'E': 0.05, 'F': 0.05, 'G': 0.07, 'H': 0.03 } <br> (This is the same as lecture slides, so check the result), 
    2. {'A': 1.0} 
* Code: `string` data type

#### Print Huffman code in the following format:


|Symbol|Code|
|:-:|:-:|
|A|	10|
|B|	11|
|C|	00|
|D|	01|


````python
def HuffmanCode(symbol_prob_dict):
    # Step 1: Sort the symbols by probability (ascending)
    symbol_prob_list = sorted(symbol_prob_dict.items(), key=lambda x: x[1])

    # Step 2: Apply the Huffman coding algorithm for two symbols at a time
    while len(symbol_prob_list) > 2:
        # Step 2.1: Pick the two smallest probabilities
        left_symbol, left_prob = symbol_prob_list.pop(0)
        right_symbol, right_prob = symbol_prob_list.pop(0)

        # Step 2.2: Merge them and assign '0' to the left and '1' to the right
        combined_prob = left_prob + right_prob
        combined_symbol = left_symbol + right_symbol

        # Step 2.3: Insert the merged symbol back into the sorted list
        symbol_prob_list.append((combined_symbol, combined_prob))
        symbol_prob_list.sort(key=lambda x: x[1])

    # Step 3: Assign codes based on the order of appearance
    # First create a list of the symbols and probabilities sorted by the final order
    final_symbols = symbol_prob_list[0][0] + symbol_prob_list[1][0]

    # Initialize the result dictionary
    symbol_code_dict = {}

    # Step 4: Assign 0 and 1 to each symbol based on final sorted order
    for i, symbol in enumerate(final_symbols):
        # '0' for the first half (lower part of the sorted list), '1' for the upper part
        symbol_code_dict[symbol] = str(i % 2)

    return symbol_code_dict


# Test the function with the provided input
input_dict = {'A': 0.1, 'B': 0.19, 'C': 0.21, 'D': 0.3, 'E': 0.05, 'F': 0.05, 'G': 0.07, 'H': 0.03}

output_dict = HuffmanCode(input_dict)

# Print the Huffman codes
for symbol, code in output_dict.items():
    print(f"{symbol}: {code}")
````

#### (2) Write a code which calculates *average code length* (08.Information Theory.pdf, p.10), its *entropy* based on the following message. 
> Message: 'ABCAACAACAACAACAAAAAAAAAAAAAAAAABAAAAAACDDDDADDAADDABCAAAAAAAAAACCAAAAAACCACACCCAAAADDAAAAAAAAAAAAAA'

#### Print the following:
1. **average code length**,
2. its **entropy**,
3. the **Huffman code table** (in the following format).
|Symbol|Code|Probability|
|:-:|:-:|:-:|
|A|1|0.62|
|B|000|0.07|
|C|01|0.21|
|D|001|0.10|

````python
import heapq
from collections import defaultdict

message = 'ABCAACAACAACAACAAAAAAAAAAAAAAAAABAAAAAACDDDDADDAADDABCAAAAAAAAAACCAAAAAACCACACCCAAAADDAAAAAAAAAAAAAA'
frequency = defaultdict(int)

for char in message:
    frequency[char] += 1

heap = [[weight, [symbol, ""]] for symbol, weight in frequency.items()]
heapq.heapify(heap)

while len(heap) > 1:
    lo = heapq.heappop(heap)
    hi = heapq.heappop(heap)
    for pair in lo[1:]:
        pair[1] = '0' + pair[1]
    for pair in hi[1:]:
        pair[1] = '1' + pair[1]
    heapq.heappush(heap, [lo[0] + hi[0]] + lo[1:] + hi[1:])

huffman_code = sorted(heap[0][1:], key=lambda p: (len(p[-1]), p))
huffman_code_table = {symbol: code for symbol, code in huffman_code}

total_symbols = sum(frequency.values())
probabilities = {symbol: frequency[symbol] / total_symbols for symbol in frequency}

average_code_length = sum(frequency[symbol] * len(code) for symbol, code in huffman_code_table.items()) / total_symbols

entropy = -sum((frequency[symbol] / total_symbols) * (len(code) / len(code)) for symbol, code in huffman_code_table.items())

print("symbol  code      probability")
for symbol, code in huffman_code_table.items():
    print(f"{symbol:<7} {code:<8} {probabilities[symbol]:.2f}")

print("\nAverage Code Length:", average_code_length)
print("Entropy:", entropy)
````

#### (3) We will see what would happen when we have a wrong statistical model (i.e. probability set) for Huffman code. Do the following steps. 

> Message, $X_1$: 'ABCAACAACAACAACAAAAAAAAAAAAAAAAABAAAAAACDDDDADDAADDABCAAAAAAAAAACCAAAAAACCACACCCAAAADDAAAAAAAAAAAAAA'
> 
> (it is the same message of problem (2))
>
> Let the true probability set of input message $X_1$ as $P_1$ and the given probability as $P_2$ (in the next table).

1. Generate a Huffman code table with the following probability:

> False probability, $P_2$, and we assume that the corresponding message is $X_2$.
|Symbol|Probability|
|:-:|:-:|
|A|0.25|
|B|0.25|
|C|0.25|
|D|0.25|

2. Based on the Huffman code table of step 1, calculate *the average code length* of the message $X_1$.

> This is the scenario that we used the wrong probability model ($P_2$) for the message $X_1$.

3. Calculate the *average code length* (of step 2) - its entropy $H(X_1$): i.e., $$L(\text{Huffman}(X_1)) - H(X_1)$$.   

5. Calculate the relative entropy $D(P_1||P_2)$ (08.Information Theory.pdf, p.46).

6. Discuss the result of step 4 and 5, what is the meaning?

````python
import heapq
from collections import defaultdict
import math

message_X1 = 'ABCAACAACAACAACAAAAAAAAAAAAAAAAABAAAAAACDDDDADDAADDABCAAAAAAAAAACCAAAAAACCACACCCAAAADDAAAAAAAAAAAAAA'
frequency_X1 = defaultdict(int)

for char in message_X1:
    frequency_X1[char] += 1

total_symbols_X1 = sum(frequency_X1.values())
P1 = {symbol: frequency_X1[symbol] / total_symbols_X1 for symbol in frequency_X1}

false_probabilities = {'A': 0.25, 'B': 0.25, 'C': 0.25, 'D': 0.25}

heap = [[weight, [symbol, ""]] for symbol, weight in false_probabilities.items()]
heapq.heapify(heap)

while len(heap) > 1:
    lo = heapq.heappop(heap)
    hi = heapq.heappop(heap)
    for pair in lo[1:]:
        pair[1] = '0' + pair[1]
    for pair in hi[1:]:
        pair[1] = '1' + pair[1]
    heapq.heappush(heap, [lo[0] + hi[0]] + lo[1:] + hi[1:])

huffman_code_table_P2 = sorted(heap[0][1:], key=lambda p: (len(p[-1]), p))
huffman_code_table_P2 = {symbol: code for symbol, code in huffman_code_table_P2}

average_code_length_P2 = sum(false_probabilities[symbol] * len(code) for symbol, code in huffman_code_table_P2.items())

entropy_P1 = -sum(P1[symbol] * math.log2(P1[symbol]) for symbol in P1)

D_P1_P2 = sum(P1[symbol] * math.log2(P1[symbol] / false_probabilities[symbol]) for symbol in P1)

print("Huffman Code Table based on False Probabilities (P2):")
for symbol, code in huffman_code_table_P2.items():
    print(f"{symbol}: {code}")

print(f"\nAverage Code Length (P2): {average_code_length_P2}")
print(f"Entropy of X1 (P1): {entropy_P1}")
print(f"Relative Entropy D(P1 || P2): {D_P1_P2}")
````
