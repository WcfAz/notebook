# Linear Programming

### Solving LP problems using Python Libraries

Now, we will solve LP problems using available Python libraries. <br>
There are many options to do it but we will focus on two packages, SciPy and PuLP. 
* <span style='color:#d63ece'> __SciPy__ </span> is a general-purpose package for scientific computing with Python.
* <span style='color:#d63ece'> __PuLP__ </span> is a Python linear programming API for defining problems and invoking external solvers.

Pros and cons
* Both packages can solve mixed integer linear programming (in case of SciPy, `scipy.optimize.milp()`)
* SciPy can solve non-linear optimization problem but PuLP (up to 3.1.1 version) cannot solve it.
* PuLP supports both min and max problems, but SciPy support only min problem.

### 1. Using SciPy

To define and solve optimization problems with SciPy, you need to import `scipy.optimize.linprog()`:
https://docs.scipy.org/doc/scipy/reference/generated/scipy.optimize.linprog.html

```python
from scipy.optimize import linprog
```

Now that you have linprog() imported, you can start optimizing.

#### Example 1
Let’s first solve the linear programming problem from above:

$ 
\text{Maximize} \quad  z = x + 2y $ <br>

$
\text{subject to:}
\begin{cases}
+2x + y &\leq 20\\
-4x + 5y &\leq 10 \\
-x + 2y &\geq -2 \\
-x + 5y &= 15 \\
\phantom{+} x &\geq 0 \\
\phantom{+} y &\geq 0
\end{cases}
$

**NOTE**: 
* linprog() solves **only minimization** (not maximization) problems,
* and _doesn’t allow inequality constraints with the greater than or equal to sign_ (≥).
To work around these issues, you need to modify your problem before starting optimization:

After introducing these changes, you get a new system:
$ 
\text{minimize} \quad  -z = -x - 2y $ <br>

$
\text{subject to:}
\begin{cases}
+2x + y &\leq 20\\
-4x + 5y &\leq 10 \\
+x - 2y &\leq +2 \\
-x + 5y &= 15 \\
\phantom{+} x &\geq 0 \\
\phantom{+} y &\geq 0
\end{cases}
$

##### The next step is to define the input values:

You put the values from the system above into the appropriate lists, tuples, or NumPy arrays:

* __obj__ holds the coefficients from the objective function.
* __lhs_ineq__ holds the left-side coefficients from the inequality constraints.
* __rhs_ineq__ holds the right-side coefficients from the inequality constraints.
* __lhs_eq__ holds the left-side coefficients from the equality constraint.
* __rhs_eq__ holds the right-side coefficients from the equality constraint.


**Note**: Please, be careful with the order of rows and columns!
* The order of the rows for the left and right sides of the constraints must be the same. Each row represents one constraint.
* The order of the coefficients from the objective function and left sides of the constraints must match. Each column corresponds to a single decision variable.



```python
obj = [-1, -2]   # coefficient for x and y

lhs_ineq = [[ 2,  1],  
            [-4,  5],  
            [ 1, -2]]  
rhs_ineq = [20,  
            10,  
             2]  
lhs_eq = [[-1, 5]]  
rhs_eq = [15]       
```

##### The next step is to define the bounds for each variable in the same order as the coefficients. 
In this case, they’re both between zero and positive infinity:

```python
import numpy as np

bnd = [(0, np.inf),  # Bounds of x
       (0, np.inf)]  # Bounds of y
```

This statement is redundant because linprog() takes these bounds (zero to positive infinity) *by default*. <br>
Finally, it’s time to optimize and solve your problem of interest. You can do that with `linprog()`:

```python
opt = linprog(c=obj, A_ub=lhs_ineq, b_ub=rhs_ineq,
              A_eq=lhs_eq, b_eq=rhs_eq, bounds=bnd,
              method="highs")
opt
```

##### OptimizeResult

A scipy.optimize.OptimizeResult consisting of the fields below. <br>
Note that the return types of the fields may depend on whether the optimization was successful, therefore it is recommended to check OptimizeResult.status before relying on the other fields:

* __x__ 1-D array: The values of the decision variables that minimizes the objective function while satisfying the constraints.
* __fun__ float: The optimal value of the objective function.
* __slack__ 1-D array: The (nominally positive) values of the slack variables.
* __con__ 1-D array: The (nominally zero) residuals of the equality constraints.
* __success__ bool: True when the algorithm succeeds in finding an optimal solution.
* __status__ int: An integer representing the exit status of the algorithm. <br>
        0 : Optimization terminated successfully. <br>
        1 : Iteration limit reached. <br>
        2 : Problem appears to be infeasible. <br>
        3 : Problem appears to be unbounded. <br>
        4 : Numerical difficulties encountered. <br>
* __nit__ int: The total number of iterations performed in all phases.


##### More on Residuals and Marginals
##### Residuals:

*    For inequality constraints (ineqlin), residuals are the values of the slack variables, indicating how much the constraint is "__slack__" or how much the current solution is below the constraint's limit.
*    For equality constraints (eqlin), residuals represent the difference between the constraint's value and the right-hand side, ideally being close to zero.
*    For bounds on decision variables, residuals show how much the variable values deviate from their lower or upper bounds. 

##### Marginals (AKA dual values / shadow prices / Lagrange multipliers):

*    Marginals are the partial derivatives of the objective function with respect to the right-hand side of the constraints.
*    They indicate how much the objective function's value would change if the right-hand side of a particular constraint were to increase by one unit, while keeping all other constraints constant.
*    For example, a positive marginal for an inequality constraint means that increasing the constraint's limit would decrease the objective function's value, and vice-versa. 

##### Example:
Imagine you have an inequality constraint $x_1 + x_2 \leq 5$ and the optimal solution is $x_1 = 3, x_2 = 1$. <br> The residual would be 1 (5 - (3 + 1)), showing that the constraint is not binding. <br> The marginal for this constraint might be -0.5, <br>
meaning that if the right-hand side (= 5) were increased by one unit, the objective function value would decrease by 0.5 units. 

In essence, residuals quantify the slack or difference from the constraint's limit, while marginals provide insight into the sensitivity of the objective function to changes in the constraints

You can access there valuse separately:

```python
print(opt.success)

print(opt.fun)

print(opt.x)
```

After installation of the PuLP, you start by importing what you need:

```python
from pulp import LpMaximize, LpProblem, LpStatus, lpSum, LpVariable
```

We will solve LP problem of __example 1__ using PuLP. <br>
$ 
\text{Maximize} \quad  z = x + 2y $ <br>

$
\text{subject to:}
\begin{cases}
+2x + y &\leq 20\\
-4x + 5y &\leq 10 \\
-x + 2y &\geq -2 \\
-x + 5y &= 15 \\
\phantom{+} x &\geq 0 \\
\phantom{+} y &\geq 0
\end{cases}
$

##### The first step is to initialize an instance of LpProblem to represent your model:

```python
# Create the model
model = LpProblem(name="Example_1", sense=LpMaximize)
```

You use the `sense` parameter to choose whether to perform _minimization_ (LpMinimize or 1, which is the default) or _maximization_ (LpMaximize or -1). <br> This choice will affect the result of your problem. 

Once that you have the model, you can define the decision variables as instances of the `LpVariable` class:

```python
# Initialize the decision variables
x = LpVariable(name="x", lowBound=0)
y = LpVariable(name="y", lowBound=0)
```

You need to provide a lower bound with `lowBound`=0 because the default value is negative infinity. <br> The parameter `upBound` defines the upper bound, but you can omit it here because it defaults to positive infinity.

The optional parameter cat defines the category of a decision variable. If you’re working with continuous variables, then you can use the default value "Continuous".

You can use the variables x and y to create other PuLP objects that represent linear expressions and constraints:

```python
expression = 2 * x + 4 * y
print(type(expression))

constraint = 2 * x + 4 * y >= 8
print(type(constraint))
```

##### The next step is to create the constraints and objective function as well as to assign them to your model. 
You don’t need to create lists or matrices. <br> Just write Python expressions and use the += operator to append them to the model:

```python
# Add the constraints to the model
model += (2 * x + y <= 20, "c1")
model += (4 * x - 5 * y >= -10, "c2")
model += (-x + 2 * y >= -2, "c3")
model += (-x + 5 * y == 15, "c4")
```

In the above code, you define tuples that hold the constraints and their names. <br> 
LpProblem allows you to add constraints to a model by specifying them as tuples. <br> 
The first element is a LpConstraint instance. The second element is a human-readable name for that constraint.

Setting the objective function is very similar:

```python
# Add the objective function to the model
obj_func = x + 2 * y
model += obj_func
```

You can now see the full definition of this model:

```python
model
```

Finally, you’re ready to solve the problem. You can do that by calling `.solve()` on your model object. <br>
If you want to use the default solver (CBC), then you don’t need to pass any arguments:

```python
# Solve the problem
status = model.solve()
```

`.solve()` calls the underlying solver, modifies the model object, and returns the integer status of the solution, which will be 1 if the optimum is found. <br> For the rest of the status codes, see `LpStatus[]`.

You can get the optimization results as the attributes of model. <br> 
The function `value()` and the corresponding method `.value()` return the actual values of the attributes:

```python
# check status
print(f"status: {model.status}, {LpStatus[model.status]}")

# check the optimal value
print(f"objective: {model.objective.value()}")

# check bfs
for var in model.variables():
    print(f"{var.name}: {var.value()}")

# check the slack variabls
for name, constraint in model.constraints.items():
    print(f"{name}: {constraint.value()}")
```

If you want to run a different solver, then you can specify it as an argument of .solve(). <br> 
For example, if you want to use GLPK and already have it installed, <br>
then you can use <mark style="background-color: lightgreen"> solver=GLPK(msg=False)</mark> in the last line. <br>
Keep in mind that you’ll also need to import it:

```python
from pulp import GLPK

# Create the model
model = LpProblem(name="Example1_GLPK", sense=LpMaximize)

# Initialize the decision variables
x = LpVariable(name="x", lowBound=0)
y = LpVariable(name="y", lowBound=0)

# Add the constraints to the model
model += (2 * x + y <= 20, "c1")
model += (4 * x - 5 * y >= -10, "c2")
model += (-x + 2 * y >= -2, "c3")
model += (-x + 5 * y == 15, "c4")

# Add the objective function to the model
model += lpSum([x, 2 * y])                      # for larger problems, `lpSum()` is more convenient.

# Solve the problem
status = model.solve(solver=GLPK(msg=False))    # <----- Here, we can specify a solver.
```

The `msg` parameter is used to display information from the solver. <br>
Your model is defined and solved, so you can inspect the results the same way you did in the previous case:

```python
print(f"status: {model.status}, {LpStatus[model.status]}")

print(f"objective: {model.objective.value()}")

for var in model.variables():
    print(f"{var.name}: {var.value()}")

for name, constraint in model.constraints.items():
    print(f"{name}: {constraint.value()}")
```

### mixed-integer linear programming
You can also use PuLP to solve mixed-integer linear programming problems. <br>
To define an integer or binary variable, just pass `cat="Integer"` or `cat="Binary"` to LpVariable. <br>
Everything else remains the same:

```python
# Create the model
model = LpProblem(name="small-problem", sense=LpMaximize)

# Initialize the decision variables: x is integer, y is continuous
x = LpVariable(name="x", lowBound=0, cat="Integer")
y = LpVariable(name="y", lowBound=0)

# Add the constraints to the model
model += (2 * x + y <= 20, "c1")
model += (4 * x - 5 * y >= -10, "c2")
model += (-x + 2 * y >= -2, "c3")
model += (-x + 5 * y == 15, "c4")

# Add the objective function to the model
model += lpSum([x, 2 * y])

# Solve the problem
status = model.solve()
```

This time, we specify the value of x as _integer_, the solution is different from the previous result. <br>
The result show that <mark> x = 7.0 </mark>, i.e., x is an integer value not continuous one.

```python
print(f"status: {model.status}, {LpStatus[model.status]}")

print(f"objective: {model.objective.value()}")

for var in model.variables():
    print(f"{var.name}: {var.value()}")

for name, constraint in model.constraints.items():
    print(f"{name}: {constraint.value()}")
```

### Visualization of LP problem
Since this LP problem contains only two variables, and we know how to draw a figure from the previous LAB, we can draw the result of LP solution graphically. <br> 
In fact, the current LP problem doesn't have a feasible region as an area but a line because one of constraint is the form of equality. <br>
So, inside of imshow(), the equality constraint is modified to an inequality one to show you an example of how to draw a feasible region. 

More on imshow() extent: <br>
https://matplotlib.org/stable/users/explain/artists/imshow_extent.html

```python
import matplotlib.pyplot as plt
import numpy as np

# model += (2 * x + y <= 20, "c1")
# model += (4 * x - 5 * y >= -10, "c2")
# model += (-x + 2 * y >= -2, "c3")
# model += (-x + 5 * y == 15, "c4")
# obj_func = x + 2 * y = 16.81817

# plot the feasible region
d = np.linspace(-3,20,500)    # you need to know the range of your graph in advance
x,y = np.meshgrid(d,d)
plt.imshow( ((x>=0) & (y>=0) 
             & (2 * x + y <= 20) 
             & (4 * x - 5 * y >= -10) 
             & (-x + 2 * y >= -2)
             & (-x + 5 * y <= 15)).astype(int) , 
           extent=(x.min(),x.max(),y.min(),y.max()),
           origin="lower", cmap="Greens", alpha = 0.1);


# plot the lines defining the constraints
x = np.linspace(-3, 20, 500)

# non-negative constraints
y1 = (x*0)   # y >= 0
x1 = 0       # x >= 0

# other constraints
y2 = (20 - 2*x)          #  y <= 20 - 2*x
y3 = (4 * x +10)/5.0    #  5 * y <= 4*x 10 
y4 = (x - 2)/2.0         #  2 * y >= x -2
y5 = (x + 15)/5.0        #  5 * y == x + 15 

# objective function
y6 = - x/2.0 + 16.81817/2.0     # 2 * y = -x + 16.81817 (this optimal value came from the above LP solution.)

# plotting
plt.plot(x, y2, 'r-', label=r'$y\leq 20-2x$')
plt.plot(x, y3, 'g-', label=r'$5y\leq -4x+10$')
plt.plot(x, y4, 'b-', label=r'$2y\geq x-2$')
plt.plot(x, y5, 'm-', label=r'$5y=x+15$')
plt.plot(x, y6, 'k:', label='obj. fun')
plt.xlim(0,10)
plt.ylim(0,10)
plt.legend(bbox_to_anchor=(1.05, 1), loc=2, borderaxespad=0.)
plt.xlabel(r'$x$')
plt.ylabel(r'$y$')
plt.grid()

# plt.savefig('fig4.png')
```

This LP problem contains one equality constraint (<span style='color:magenta'>magenta color</span>), so the optimal solution must be found on the equality constraint line.
