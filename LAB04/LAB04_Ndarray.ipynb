#### (1) Generate two matrices A, B and perform matrix multiplication and Hadamard product.

$ A =

#### (5) Generate a 4x4 matrix which includes sequential integers from 3 to 19 randomly. In the matrix, if the element is a prime number, then add 100. 

#### Example
$ M =
     \begin{bmatrix} 6 & 9 & 3 & 7 \\ 5 & 8 & 12 & 16 \\ 3 & 19 & 12 & 11 \\ 4 & 14 & 9 & 13 \end{bmatrix} $
     
#### After running your code
$ M_{new} =
     \begin{bmatrix} 6 & 9 & 103 & 107 \\ 105 & 8 & 12 & 16 \\ 103 & 119 & 12 & 111 \\ 4 & 14 & 9 & 113 \end{bmatrix} $


````python
import numpy as np
def is_prime(n):
    if n <= 1:
        return False  
    for i in range(2, int(n**0.5) + 1): 
        if n % i == 0:  
            return False
    return True  
matrix = np.random.randint(3, 19, size=(4, 4))
print(matrix)
for i in range(4):
    for j in range(4):
        if is_prime(matrix[i, j]):
            matrix[i, j] += 100
print(matrix)
````
