#### (1) Generate message #1 with the following conditions. Caluculate probability and entropy of the message. Also, draw probability graph with bar plot.
#### Condition: 
__message #1__
* Containing 4 letters A, B, C, D
* Given probability set = [0.25, 0.25, 0.25, 0.25] for generating message.
* length = 10000
* Random letters
* __Calculate its entropy using counted probability__ of the generated message.

__Note__ : Since the message is generated randomly, its letter's probability can not be the same as given probability set <br>  e.g., P{A} = 0.25 can be P{A} = 0.249 or 0.251, and so on.

````python
import numpy as np
import matplotlib.pyplot as plt
from collections import Counter

length = 10000
probabilities = [0.25, 0.25, 0.25, 0.25]  

letters = ['A', 'B', 'C', 'D']

message = np.random.choice(letters, size=length, p=probabilities)

letter_counts = Counter(message)

generated_probabilities = [letter_counts[letter] / length for letter in letters]

entropy = -sum(p * np.log2(p) for p in generated_probabilities if p > 0)

plt.bar(letters, generated_probabilities, color='skyblue')
plt.xlabel('Letters')
plt.ylabel('Probability')
plt.title('Probability Distribution of Generated Message')
plt.show()

print(f"Entropy: {entropy}")
print(f"Generated Probabilities: {generated_probabilities}")
````

#### (5) Draw probability graph of message #1 to #4 using 2x2 subplot and bar plot.

````python
import numpy as np
import matplotlib.pyplot as plt
from collections import Counter

lengths = [10000, 10000, 10000, 10000]
probabilities_list = [
    [0.25, 0.25, 0.25, 0.25], 
    [0.40, 0.30, 0.20, 0.10],  
    [0.25, 0.25, 0.40, 0.10],  
    [0.25, 0.25, 0.40, 0.10]  
]

letters = ['A', 'B', 'C', 'D']

fig, axes = plt.subplots(2, 2, figsize=(10, 8))

for i in range(4):
    message = np.random.choice(letters, size=lengths[i], p=probabilities_list[i])
    letter_counts = Counter(message)
    generated_probabilities = [letter_counts[letter] / lengths[i] for letter in letters]

    entropy = -sum(p * np.log2(p) for p in generated_probabilities if p > 0)

    ax = axes[i // 2, i % 2]
    ax.bar(letters, generated_probabilities, color='skyblue')
    ax.set_title(f"Message #{i+1}\nEntropy: {entropy:.2f}")
    ax.set_xlabel('Letters')
    ax.set_ylabel('Probability')

plt.tight_layout()
plt.show()
````

#### (6) Draw entropy graph or message #1 to #4 using bar plot and discuss why the entropy of each message changes
__NOTE__ : when you calculate entropy and don't forget to avoid $log_2( 0 )$ = - infinity

````python
import numpy as np
import matplotlib.pyplot as plt

lengths = [10000, 10000, 10000, 10000]
probabilities_list = [
    [0.25, 0.25, 0.25, 0.25], 
    [0.40, 0.30, 0.20, 0.10],  
    [0.25, 0.25, 0.40, 0.10],  
    [0.25, 0.25, 0.40, 0.10]  
]

letters = ['A', 'B', 'C', 'D']

def calculate_entropy(probabilities):
    return -sum(p * np.log2(p) for p in probabilities if p > 0)

entropies = []

for probs in probabilities_list:
    entropy = calculate_entropy(probs)
    entropies.append(entropy)

plt.bar([f'Message #{i+1}' for i in range(4)], entropies, color='skyblue')
plt.xlabel('Messages')
plt.ylabel('Entropy')
plt.title('Entropy of Messages #1 to #4')
plt.show()
````

#### (7) Predicting Travel Behavior 

**Discription of the problem** 

We are given two variables related to travel behavior:
1. **X (Mode of Transport)**: This variable represents the mode of transport chosen by the individual. It has three possible values:
   - $ X_1 $: "Car"
   - $ X_2 $: "Bus"
   - $ X_3 $: "Bike"

2. **Y (Trip Purpose)**: This variable represents the purpose of the trip. It also has three possible values:
   - $ Y_1 $: "Work"
   - $ Y_2 $: "School"
   - $ Y_3 $: "Leisure"

### Joint Probability Distribution

The **joint probability distribution** $ P(X, Y) $ tells us the probability of each combination of mode of transport and trip purpose. The table below provides the joint probabilities for each combination:

| Mode of Transport (X) | Trip Purpose (Y) | $ P(X, Y) $ |
|-----------------------|------------------|---------------|
| Car                   | Work             | 0.30          |
| Car                   | School           | 0.05          |
| Car                   | Leisure          | 0.15          |
| Bus                   | Work             | 0.10          |
| Bus                   | School           | 0.10          |
| Bus                   | Leisure          | 0.05          |
| Bike                  | Work             | 0.05          |
| Bike                  | School           | 0.10          |
| Bike                  | Leisure          | 0.10          |


This table tells us the likelihood of each combination of **mode of transport** and **trip purpose**. For example:
- The probability that someone uses a **Car** for **Work** is $ P(X = \text{Car}, Y = \text{Work}) = 0.30 $.
- The probability that someone uses a **Bus** for **Leisure** is $ P(X = \text{Bus}, Y = \text{Leisure}) = 0.05 $.


**Write a code to calculate the following entropy**:

1. **Entropy of X ( H(X) )**: Measures the uncertainty or unpredictability of the mode of transport.
2. **Entropy of Y ( H(Y) )**: Measures the uncertainty or unpredictability of the trip purpose.
3. **Joint Entropy of X and Y ( H(X,Y) )**: Measures the uncertainty of both variables (mode of transport and trip purpose) together.
4. **Conditional Entropy ( H(Y|X) and H(X|Y) )**: Measures the uncertainty in one variable given the other.
5. **Mutual Information (I(X;Y))**: Measures how much knowing one variable reduces the uncertainty of the other variable.
