## &#9724; Sets

Sets are mainly used to eliminate repeated numbers in a sequence/list. It is also used to perform some standard set operations.

Sets are declared as set() which will initialize a empty set. Also `set([sequence])` can be executed to declare a set with elements

````python
set1 = set()
print(type(set1))
print(set1)
````

````python
set0 = set([1,2,2,3,3,4])
set0 = {1,2,2,3,3,4} # equivalent to the above
print(set0)
````

#### Python Set methods


* __add()__: 	Adds an element to the set
* __clear()__: 	Removes all elements from the set
* __copy()__: 	Returns a copy of the set
* __difference()__: 	Returns the difference of two or more sets as a new set
* __difference_update()__: 	Removes all elements of another set from this set
* __discard()__: 	Removes an element from the set if it is a member. (Do nothing if the element is not in set)
* __intersection()__: 	Returns the intersection of two sets as a new set
* __intersection_update()__: 	Updates the set with the intersection of itself and another
* __isdisjoint()__: 	Returns True if two sets have a null intersection
* __issubset()__: 	Returns True if another set contains this set
* __issuperset()__: 	Returns True if this set contains another set
* __pop()__: 	Removes and returns an arbitary set element. Raise KeyError if the set is empty
* __remove()__: 	Removes an element from the set. If the element is not a member, raise a KeyError
* __symmetric_difference()__: 	Returns the symmetric difference of two sets as a new set
* __symmetric_difference_update()__: 	Updates a set with the symmetric difference of itself and another
* __union()__: 	Returns the union of sets in a new set
* __update()__: 	Updates the set with the union of itself and others

````python
set2 = set([2,3,4,5])
````

Dictionaries are mappings between keys and items stored in the dictionaries. Alternatively one can think of dictionaries as sets in which something stored against every element of the set. They can be defined as follows:

To define a dictionary, equate a variable to { } or dict()

````python
d = dict() # or equivalently d={}
print(type(d))
d['abc'] = 3
d[4] = "A string"
print(d)

d['abc']
````

As can be guessed from the output above. Dictionaries can be defined by using the `{ key : value }` syntax. The following dictionary has three elements

````python
d = { 1: 'One', 2 : 'Two', 100 : 'Hundred'}
len(d)
````

Now you are able to access 'One' by the index value set at 1

````python
print(d[1])
````

There are a number of alternative ways for specifying a dictionary including as a list of `(key,value)` tuples.
To illustrate this we will start with two lists and form a set of tuples from them using the **zip()** function
Two lists which are related can be merged to form a dictionary.

````python
names = ['One', 'Two', 'Three', 'Four', 'Five']
numbers = [1, 2, 3, 4, 5]

[(name,number) for name,number in zip(names,numbers)] # create (name,number) pairs
````

````python
for item in zip(names, numbers):
    print(item)

type(item)
````

#### Dictionary Comprehension
Now we can create a dictionary that maps the name to the number as follows.

````python
a1 = dict((name,number) for name,number in zip(names,numbers)) # dictionary
print(a1)
print(type(a1))
print()

a2 = {(name,number) for name,number in zip(names,numbers)} # set
print(a2)
print(type(a2))
print()

a3 = {name : len(name) for name in names}
print(a3)
print(type(a3))
````

Note that the ordering for this dictionary is not based on the order in which elements are added but on its own ordering (based on hash index ordering). It is best never to assume an ordering when iterating over elements of a dictionary.

By using tuples as indexes we make a dictionary behave like a sparse matrix:

````python
matrix={(0,1): 3.5, (2,17): 0.1}
matrix[2,2] = matrix[0,1] + matrix[2,17]
print(matrix)
````

#### Example 1 : Binary string to binary list

````python
def BinStr2BinList(inputValue):
    output = []     # empty list
    Len = len(inputValue)
    # string to list
    for i in range(Len): 
        output.append(int(inputValue[i]))
    return output

# Python script
strVal = '1101'
outVal = BinStr2BinList(strVal)
print(outVal)
````

Alternatively, we can do the same thing using <font color=blue> *list comprehension* </font> instead of a `function` definition.

````python
# Binary string to binary list
strVal = '1101'
outVal = list(int(i) for i in strVal)
print(outVal)
````

This is also possible but the list contains strings instead of int.

````python
# Binary string to binary list
strVal = '1101'
outVal = list(strVal)
print(outVal)
````

#### Example 2: Decimal to binary number ( `list` output )

$4_{10} = 100_{2}$   

4/2 $\to$ q: 2,   r: 0  
2/2 $\to$ q: 1,   r: 0  
1/2 $\to$ q: 0,   r: 1  


So, $100_{2}$ , i.e., we need to reverse the remainders.

* dec_2_bin_list()
* var_bin

````python
def Dec2BinL(value):
    record = []
    quotient = value
    while (quotient != 0):
        remainder = quotient%2
        quotient = quotient//2
        record.append(remainder)
    
    if not record: # if the list record is empty, i.e., value = 0
        record = [value]
    return record[::-1]

# Python script
print(Dec2BinL(0))
print(Dec2BinL(10))
print(Dec2BinL(13))
print(Dec2BinL(21))
````

#### Example 3: with some reversing list options

````python
def Dec2BinLOpt(value, opt):
    record = []
    quotient = value
    while (quotient != 0):
        remainder = quotient%2
        quotient = quotient//2
        record.append(remainder)
    
    if not record: # if the list record is empty, i.e., value = 0
        record = [value]
    
    # The below shows different way to reverse your list
    if opt == 1:
        out_val = record[::-1]           # This is one option to reverse list
    elif opt == 2:
        out_val = list(reversed(record)) # this can be the 2nd option
    else:
        record.reverse()                 # this is another option
        out_val = record
        
    return out_val

# Python script
print(Dec2BinLOpt(0,1))
print(Dec2BinLOpt(10,2))
print(Dec2BinLOpt(13,3))
print(Dec2BinLOpt(21,1))
````

#### Example 4: Decimal to binary number ( `string` output ) 

````python
def Dec2BinS(value):
    output = ''        # empty string
    while (value != 0):
        remainder = value%2
        value = value//2
        output = str(remainder) + output
        
    if not output:     # if string output is empty, i.e., value = 0
        output = str(value)
    return output

# Python script
print(Dec2BinS(0))
print(Dec2BinS(10))
print(Dec2BinS(13))
print(Dec2BinS(21))
````

#### Example 6: Decimal to binary number (another option)

````python
import math # if you import math module, you can use all functions in math module

def Dec2BinA(value):
    bitLen = math.floor(math.log2(value+1)+1) # ceil() and log2() are functions in math module
    output = ''
    for i in range(bitLen):
        output = str(value%2) + output
        value = value >> 1
        
    return output

# Python script
print(Dec2BinA(0))
print(Dec2BinA(10))
print(Dec2BinA(13))
print(Dec2BinA(16))
print(Dec2BinA(21))
````

### Logic and shift operators

#### AND

````python
a = 0b1010
b = 0b1111
c = a & b
print('a = {}, b = {}, c = {}\n'.format(a, b, c))
print('a: {:>7}\nb: {:>7}\nc: {:>7}'.format(bin(a), bin(b), bin(c)))
````

#### OR

````python
a = 0b1010
b = 0b1111
c = a | b
print('a = {}, b = {}, c = {}\n'.format(a, b, c))
print('a: {:>7}\nb: {:>7}\nc: {:>7}'.format(bin(a), bin(b), bin(c)))
````

#### XOR

````python
a = 0b1010
b = 0b1111
c = a ^ b
print('a = {}, b = {}, c = {}\n'.format(a, b, c))
print('a: {:>7}\nb: {:>7}\nc: {:>7}'.format(bin(a), bin(b), bin(c)))
````

#### NOT: filps each bit (e.g, 1 to 0, or 0 to 1), but it returns two's complement notation

1101: org <br>
0010: flipped <br>
1101+1: 2's complement <br>
1110: final NOT operator's result

````python
a = 0b1101
b = ~a
print('a = {}, b = {}\n'.format(a, b))
print('a: {:>7}\nb: {:>7}'.format(bin(a), bin(b)))
````

#### Shift (logical shift)

````python
a = 0b1010
b = a << 2
c = a >> 1
print('a = {}, b = {}, c = {}\n'.format(a, b, c))
print('a: {:>7}\nb: {:>7}\nc: {:>7}'.format(bin(a), bin(b), bin(c)))
````
