#### (6) Write a Python program to make all values of the dictionary in problem (5) upper cases (i.e., 'a' -> 'A'). Print the result.


```python
d={1:'a', 2:'b', 3:'c', 4:'d', 5:'e', 6:'f'}
for key in d:
    d[key]=d[key].upper()
print(d)
```

#### (7) Write a Python program to remove duplicates (i.e., if there is any same value) from the following dictionary.
**Given dictionary** = {4:4, 1:1, 5:1, 2:2, 3:4} <br>
There are two 1s and two 4s. These values are duplicated.<br>
Expected result:<br>
{4: 4, 1: 1, 2: 2}

```python
givenDict = {4: 4, 1: 1, 5: 1, 2: 2, 3: 4}
values = set()
resultDict = {}
for key, value in givenDict.items():
    if value not in values:
        resultDict[key] = value
        values.add(value)
print(resultDict)
```

#### (8) Write a Python program to combine two dictionary adding values for common keys. Print the result.
hw  = {'a':100, 'b':90, 'c':75, 'd':80} <br>
lab = {'a':90, 'b':100, 'c':85, 'd':90} <br>

ex) <br>
sumDict = {'a':190, 'b':190, 'c':160, 'd':170}

```python
hw = {'a':100, 'b':90, 'c':75, 'd':80}
lab = {'a':90, 'b':100, 'c':85, 'd':90}
sumDict={}
for key in hw:
    if key in lab:
        sumDict[key]=hw[key]+lab[key]
print("sumDict=",sumDict)
```

#### (10) Write a Python program to print all unique values in the following dictionary.
exDict = {1:1, 2:1, 3:3, 4:2, 5:7, 6:18, 7:7, 8:14, 9:20, 10:14}

```python
exDict = {1:1, 2:1, 3:3, 4:2, 5:7, 6:18, 7:7, 8:14, 9:20, 10:14}
uniqueValue = [value for value in exDict.values() if list(exDict.values()).count(value) == 1]
print("Unique values:", uniqueValue)
```
