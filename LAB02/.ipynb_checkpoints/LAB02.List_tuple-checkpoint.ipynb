#### (9) Write a Python program to convert a list of characters into a string.
**Given list**: ['D', 'i', 's', 'c', 'r', 'e', 't', 'e', ' ', 'M', 'a', 't', 'h']


````python
alist = ['D', 'i', 's', 'c', 'r', 'e', 't', 'e', ' ', 'M', 'a', 't', 'h']
result_string = ""
for char in alist:
    result_string += char
print("The converted string is:", result_string)
````

#### (10) Write a Python program to create a list by concatenating a given list which range goes from 1 to n.
**For example:** <br>
**Given list** : ['p', 'q'] <br>
**n = 10** <br>
**Expected Output** : ['p1', 'q1', 'p2', 'q2', 'p3', 'q3', 'p4', 'q4', 'p5', 'q5', ..., 'q10']

NOTE: `str()` changes an integer to a string.

````python
char_list = ['p', 'q']
n = 10
result_list = []
for i in range(1, n + 1):
    for char in char_list:
        result_list.append(char + str(i))
print("The concatenated list is:", result_list)
````

#### (11) Write a Python program to convert a tuple to a string.

**Given tuple**: ('M', 'a', 't', 'h')

````python
tup=('M', 'a', 't', 'h')
result_string = ""
for char in tup:
    result_string += char
print("Converted string:", result_string)
````

#### (15) Write a Python program to replace the last value of tuples in a list.

**Given list**: [(8, 4, 12), (4, 8, 12), (16, 2, 18), (1, 32, 24)] <br>
**New value**: 32 <br>
**Expected output**: [(8, 4, 32), (4, 8, 32), ...]

````python
lista = [(8, 4, 12), (4, 8, 12), (16, 2, 18), (1, 32, 24)]
new_value = 32
listb = [tup[:-1] + (new_value,) for tup in lista]
print("List after replacing last values:", listb)
````
