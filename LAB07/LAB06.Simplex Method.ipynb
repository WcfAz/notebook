#### 1. Write a pivot operation function (Refer to lecture notebook 6 for more details).

**Condition**:
* input: Matrix A in numpy ndarray

$ A = 
  \begin{bmatrix} 1 & 4 & 2 & 6 \\ 3 & 14 & 8 & 16\\
                  4 & 21 & 10 & 28 \end{bmatrix} $
                  
* output: Resulting matrix in numpy ndarray

````python
import numpy as np


A = np.array([[1, 4, 2, 6],
              [3, 14, 8, 16],
              [4, 21, 10, 28]], dtype=float)


def pivot(A):
   
    n = A.shape[0]
    
    
    for i in range(n):
        
        max_row = np.argmax(np.abs(A[i:n, i])) + i  
        if i != max_row:
            
            A[[i, max_row]] = A[[max_row, i]]
        
       
        for j in range(i + 1, n):
            factor = A[j, i] / A[i, i]
            A[j, i:] -= factor * A[i, i:]
    
    return A


A_pivoted = pivot(A)
print("Pivoted Matrix:\n", A_pivoted)
````

#### 2. Write a modified pivot operation function with a target term (Refer to lecture notebook 6 for more details). 

**Condition**:
* Input: Matrix A in numpy ndarray and target term in tuple (0, 2) after that, 1st result with target (1, 1).
         
$ A = 
  \begin{bmatrix} -6 & 0 & 1 & -2 & 2 & 6 \\ 3 & 1 & -1 & 8 & 1 & 9\\
                  -4 & 1 & 1 & 7 & 3 & 0 \end{bmatrix} $
                  
* Output: Resulting matrix in numpy ndarray.
* Let's say your function is `PivotWithTarget()`, then you will run your code like this. .

````python
import numpy as np

def PivotWithTarget(A, target):
   
    A = A.copy()

    row, col = target
    
    if A[row, col] == 0:
        for i in range(row + 1, A.shape[0]):
            if A[i, col] != 0:
                # Swap rows
                A[[row, i]] = A[[i, row]]
                break
    
    pivot_value = A[row, col]
    for i in range(row + 1, A.shape[0]):
        if A[i, col] != 0:
            factor = A[i, col] / pivot_value
            A[i, col:] -= factor * A[row, col:]
    
    return A


A = np.array([[-6, 0, 1, -2, 2, 6], 
              [3, 1, -1, 8, 1, 9], 
              [-4, 1, 1, 7, 3, 0]], dtype=float)


print("Original Matrix:")
print(A)


target = (0, 2)
B = PivotWithTarget(A, target)
print("\nAfter pivoting with target (0, 2):")
print(B)


target = (1, 1)
C = PivotWithTarget(B, target)
print("\nAfter pivoting with target (1, 1):")
print(C)
````
