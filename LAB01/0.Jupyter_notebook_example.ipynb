### plotting example

```python
%matplotlib inline

import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 6*np.pi, 100)
plt.plot(x, np.sin(x))
plt.grid()
plt.title('sin(x) graph')
plt.show()
```

### "is" and "==" operator
- is: identity test (same object?)
- ==: equality test (same value?)

```python
'Apple' is 'Apple pie'
```

```python
'Apple' == 'Apple pie'
```

```python
'Apple' is not 'Orange'
```

```python
'APPLE'.lower()
```

```python
'APPLE'.lower() is 'apple'
```

```python
'APPLE'.lower() == 'apple'
```

```python
print('APPLE'.lower())
```

```python
x = 3
y = 3.0
print(x == y)
print(x is y)
```

```python
A = 'apple'
B = 'APPLE'.lower()
print(A is B)
```

```python
print('apple'.capitalize())
print(A.capitalize())
```

### Simple calculation: Area or a circle
* area = $\pi r^2$

```python
pi = 3.14159
radius = 2.
# area of circle
area = pi*(radius**2)
print(area)
```
