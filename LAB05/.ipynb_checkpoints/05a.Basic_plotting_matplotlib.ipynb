#### Before we draw a graph, let's check why numpy is useful for maths related programming!
How can we calculate $ cos(x) $ using list? when $ x = \{0, 1, 2, ..., 9\} $

Can we use `math.cos()` directly to our list?

```python
import math

x = list(range(10))
print(x)
y = math.cos(x)
```

#### Obviously, `math.cos()` get a number as its input, not list!
#### Then we need a few ways to handle this
(1) List comprehension

```python
import math

x = list(range(10))
print(x)
y = list(math.cos(i) for i in x)
print(y)
```

(2) Using map()

```python
import math

x = list(range(10))
print(x)
y2 = list(map(math.cos, x))
print(y2)
```

#### If you want to control the grid lines, it gets a little complicated (Object-Oriented, or OO style).

```python
import matplotlib.pyplot as plt
import numpy as np

# sine wave
x = np.arange(0,20,.01)
y = np.sin(x)

fig = plt.figure()                          # generating figure, and get figure obj
ax = fig.gca()                              # from figure obj, get axis obj  

ax.set_xticks(np.arange(0, 20.1, 3.14))     # setting x axis grid lines
ax.set_yticks(np.arange(-1.5, 1.5, 0.5))    # setting y axis grid lines

plt.plot(x,y)
plt.grid()
plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors


def plot_colortable(colors, title, sort_colors=True, emptycols=0):

    cell_width = 212
    cell_height = 22
    swatch_width = 48
    margin = 12
    topmargin = 40

    # Sort colors by hue, saturation, value and name.
    if sort_colors is True:
        by_hsv = sorted((tuple(mcolors.rgb_to_hsv(mcolors.to_rgb(color))),
                         name)
                        for name, color in colors.items())
        names = [name for hsv, name in by_hsv]
    else:
        names = list(colors)

    n = len(names)
    ncols = 4 - emptycols
    nrows = n // ncols + int(n % ncols > 0)

    width = cell_width * 4 + 2 * margin
    height = cell_height * nrows + margin + topmargin
    dpi = 72

    fig, ax = plt.subplots(figsize=(width / dpi, height / dpi), dpi=dpi)
    fig.subplots_adjust(margin/width, margin/height,
                        (width-margin)/width, (height-topmargin)/height)
    ax.set_xlim(0, cell_width * 4)
    ax.set_ylim(cell_height * (nrows-0.5), -cell_height/2.)
    ax.yaxis.set_visible(False)
    ax.xaxis.set_visible(False)
    ax.set_axis_off()
    ax.set_title(title, fontsize=24, loc="left", pad=10)

    for i, name in enumerate(names):
        row = i % nrows
        col = i // nrows
        y = row * cell_height

        swatch_start_x = cell_width * col
        swatch_end_x = cell_width * col + swatch_width
        text_pos_x = cell_width * col + swatch_width + 7

        ax.text(text_pos_x, y, name, fontsize=14,
                horizontalalignment='left',
                verticalalignment='center')

        ax.hlines(y, swatch_start_x, swatch_end_x,
                  color=colors[name], linewidth=18)

    return fig

plot_colortable(mcolors.BASE_COLORS, "Base Colors",
                sort_colors=False, emptycols=1)
plot_colortable(mcolors.TABLEAU_COLORS, "Tableau Palette",
                sort_colors=False, emptycols=2)

#sphinx_gallery_thumbnail_number = 3
plot_colortable(mcolors.CSS4_COLORS, "CSS Colors")

# Optionally plot the XKCD colors (Caution: will produce large figure)
#xkcd_fig = plot_colortable(mcolors.XKCD_COLORS, "XKCD Colors")
#xkcd_fig.savefig("XKCD_Colors.png")

plt.show()
```

#### Want to see multiple lines together?
#### Plot lines first and call "show( )" later

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(0,20,.01)
y = np.sin(x)     # sine wave
z = np.cos(x)     # new cosine wave     

fig = plt.figure()                          # generating figure, and get figure obj
ax = fig.gca()                              # from figure obj, get axis obj  

ax.set_xticks(np.arange(0, 20.1, 3.14))     # setting x axis grid lines
ax.set_yticks(np.arange(-1.5, 1.5, 0.5))    # setting y axis grid lines

plt.plot(x, y, 'g-') # this plots sine wave
plt.plot(x, z, 'r-') # and this one plots cosine wave
plt.grid()
plt.show()
```

#### In this case, we don't know which is which, so let's add legend

```python
%matplotlib inline

import matplotlib.pyplot as plt
import numpy as np

x = np.arange(0,20,.01)
y = np.sin(x)     # sine wave
z = np.cos(x)     # new cosine wave     

fig = plt.figure()                          # generating figure, and get figure obj
ax = fig.gca()                              # from figure obj, get axis obj  

ax.set_xticks(np.arange(0, 20.1, 3.14))     # setting x axis grid lines
ax.set_yticks(np.arange(-1.5, 1.5, 0.5))    # setting y axis grid lines

plt.plot(x, y)        # this plots sine wave
plt.plot(x, z)        # and this one plots cosine wave
plt.legend(('sin(x)', 'cos(x)'), loc='upper right') # loc=1 also works

plt.grid()
plt.show()
```

#### Anothe way to add legend

```python
%matplotlib inline

import matplotlib.pyplot as plt
import numpy as np

x = np.arange(0,20,.01)
y = np.sin(x)     # sine wave
z = np.cos(x)     # new cosine wave     

fig = plt.figure()                          # generating figure, and get figure obj
ax = fig.gca()                              # from figure obj, get axis obj  

ax.set_xticks(np.arange(0, 20.1, 3.14))     # setting x axis grid lines
ax.set_yticks(np.arange(-1.5, 1.5, 0.5))    # setting y axis grid lines

plt.plot(x, y, label='sin(x)')        # this plots sine wave 
plt.plot(x, z, label='cos(x)')        # and each line is tagged with each label
plt.legend(loc='upper right')

plt.grid()
plt.show()
```

#### Figure size, saving figure as a file

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(0,20,.01)
y = np.sin(x)     # sine wave
z = np.cos(x)     # new cosine wave     

fig = plt.figure(figsize = (10,4))          # figure size can be changed here, this is 10 x 4 inch
ax = fig.gca()                                

ax.set_xticks(np.arange(0, 20.1, 3.14))     
ax.set_yticks(np.arange(-1.5, 1.5, 0.5))    

plt.plot(x, y, label='sin(x)')        
plt.plot(x, z, label='cos(x)')        
plt.legend(loc='upper right')

plt.title('sin(x) and cos(x)') # Another way to create legend
plt.xlabel('time[second]')
plt.ylabel('magnitude')

plt.grid()

plt.savefig('sinuoid.png', dpi=300, transparent=False) # save figure as PNG image format

plt.show()
```

#### Customized tick labels

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(0,20,.01)
y = np.sin(x)     # sine wave
z = np.cos(x)     # new cosine wave     

fig = plt.figure(figsize = (10,4))          # figure size can be changed here, this is 10 x 4 inch
ax = fig.gca()                                

ax.set_xticks(np.arange(0, 20.1, 3.14))     
ax.set_xticklabels(['0', r'$\pi$', r'$2\pi$', r'$3\pi$', r'$4\pi$', r'$5\pi$', r'$6\pi$'], fontsize=12)
ax.set_yticks(np.arange(-1.5, 1.5, 0.5))    

plt.plot(x, y, label='sin(x)')        
plt.plot(x, z, label='cos(x)')        
plt.legend(loc='upper right')

plt.title('sin(x) and cos(x)') # Another way to create legend
plt.xlabel(r'angle($\theta$)', fontsize=12)
plt.ylabel('magnitude', fontsize=12)

plt.grid()

plt.savefig('sinuoid.png', dpi=300, transparent=False) # save figure as PNG image format

plt.show()
```

#### Multiple plots together! subplot( )

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(0,20,0.25)                     # let's use a sparse step
y = np.sin(x)     # sine wave
z = np.cos(x)     # new cosine wave     

fig = plt.figure(figsize = (10,8))          # need to control figure size a little larger
ax = fig.gca()                                

plt.subplot(211)                            # in 2x1 frame, 1st figure
plt.plot(x, y, label='sin(x)')        
plt.title('sin(x)')
plt.grid()

plt.subplot(212)                            # in 2x1 frame, 2nd figure
plt.plot(x, z, 'o', label='cos(x)')         # only dots instead of line
plt.title('cos(x)')
plt.grid()


plt.show()
```

#### Alternatively, we can use obj's

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(0,20,0.25)                     # let's use a sparse step
y = np.sin(x)     # sine wave
z = np.cos(x)     # new cosine wave     

fig, ax= plt.subplots(2, 1, figsize = (8,10)) # using subplots and get fig, axes objects

ax[0].plot(x, y, label='sin(x)')
ax[0].plot(x, y, 'r*')
ax[0].set_title('sin(x)')
ax[0].grid()
ax[0].set_xticks(np.arange(0, 20.1, 3.14))     
ax[0].set_xticklabels(['0', r'$\pi$', r'$2\pi$', r'$3\pi$', r'$4\pi$', r'$5\pi$', r'$6\pi$'], fontsize=12)
ax[0].set_yticks(np.arange(-1.5, 1.5, 0.5))
ax[0].set_xlim([0, 6*np.pi])
ax[0].set_ylim([-1.1, 1.1])
ax[0].set_xlabel('X'); ax[0].set_ylabel('Y')


ax[1].plot(x, z, 'o', label='cos(x)')         # only dots instead of line
ax[1].set_title('cos(x)')
ax[1].grid()
ax[1].set_xticks(np.arange(0, 20.1, 3.14))     
ax[1].set_xticklabels(['0', r'$\pi$', r'$2\pi$', r'$3\pi$', r'$4\pi$', r'$5\pi$', r'$6\pi$'], fontsize=12)
ax[1].set_yticks(np.arange(-1.5, 1.5, 0.5))
ax[1].set_xlim([0, 6*np.pi])
ax[1].set_ylim([-1.1, 1.1])
ax[1].set_xlabel('X'); ax[1].set_ylabel('Y');

plt.show()
```

#### Line and marker styles

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 5, 10)

fig, ax = plt.subplots(figsize=(12,6))

ax.plot(x, x+1, color="blue", linewidth=0.25)
ax.plot(x, x+2, color="blue", linewidth=0.50)
ax.plot(x, x+3, color="blue", linewidth=1.00)
ax.plot(x, x+4, color="blue", linewidth=2.00)

# possible linestype options ‘-‘, ‘--’, ‘-.’, ‘:’, ‘steps’
ax.plot(x, x+5, color="red", lw=2, linestyle='-')
ax.plot(x, x+6, color="red", lw=2, ls='-.')
ax.plot(x, x+7, color="red", lw=2, ls=':')

# custom dash
line, = ax.plot(x, x+8, color="black", lw=1.50)
line.set_dashes([5, 10, 15, 10]) # format: line length, space length, ...

# possible marker symbols: marker = '+', 'o', '*', 's', ',', '.', '1', '2', '3', '4', ...
ax.plot(x, x+ 9, color="green", lw=2, ls='--', marker='+')
ax.plot(x, x+10, color="green", lw=2, ls='--', marker='o')
ax.plot(x, x+11, color="green", lw=2, ls='--', marker='s')
ax.plot(x, x+12, color="green", lw=2, ls='--', marker='1')

# marker size and color
ax.plot(x, x+13, color="purple", lw=1, ls='-', marker='o', markersize=2)
ax.plot(x, x+14, color="purple", lw=1, ls='-', marker='o', markersize=4)
ax.plot(x, x+15, color="purple", lw=1, ls='-', marker='o', markersize=8, markerfacecolor="red")
ax.plot(x, x+16, color="purple", lw=1, ls='-', marker='s', markersize=8, 
        markerfacecolor="yellow", markeredgewidth=2, markeredgecolor="blue");
```

#### Twin axes

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 5, 100)

fig, ax1 = plt.subplots()

ax1.plot(x, x**2, color='blue')
ax1.set_ylabel(r"area $(m^2)$", fontsize=18, color="blue")
for label in ax1.get_yticklabels():
    label.set_color("blue")
    
ax2 = ax1.twinx()

ax2.plot(x, x**3, color='red')
ax2.set_ylabel(r"volume $(m^3)$", fontsize=18, color="red")
for label in ax2.get_yticklabels():
    label.set_color("red")
```

#### When we need axes at origin, i.e., (0, 0) at the center of graph

Spines are the lines connecting the axis tick marks and noting the boundaries of the data area. 

```python
import matplotlib.pyplot as plt
import numpy as np

fig, ax = plt.subplots()

ax.spines['right'].set_color('none') # the spine means the x or y axis.
ax.spines['top'].set_color('none')

ax.xaxis.set_ticks_position('bottom')
ax.spines['bottom'].set_position(('data',0)) # set position of x spine to x=0

ax.yaxis.set_ticks_position('left')
ax.spines['left'].set_position(('data',0))   # set position of y spine to y=0

x = np.linspace(-5., 5., 100)
ax.plot(x, x**3);
```

#### Simple bar plot example
plot() and bar() plot do not have any difference except the shape of their graph.

```python
%matplotlib inline

import numpy as np
import matplotlib.pyplot as plt

fig = plt.figure(figsize = (8,5))
occurList = [10, 12, 9, 8]
letterList = ['A', 'B', 'C', 'D']
rng = range(len(occurList))

plt.bar(rng, occurList)
plt.xlabel('Letter'); plt.ylabel('Occurence')
plt.xticks(rng, letterList)
tmpString = 'Occurrence of Letters from {} to {}'.format(letterList[0], letterList[len(letterList)-1])
plt.title(tmpString)
plt.show()
```
