#### (1) Draw the graph of the following equations according to the given conditions.

$ y = x\\
  y = x^2\\
  y = x^3 $

#### Condition:
* X Range: 0 to 5.
* X ticks: Every 1 (e.g., 0, 1, 2, etc.).
* Color: Red, Green, and Blue each.
* Line: Solid, Dashed, and Dotted each. 
* Title: Simple Plot.
* Legend: The equation itself for each (e.g., y=x).
* There should be grid.
* Lable: X and Y


````python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(0,10,1)
y1 = x     
y2 = x*x      
y3 = x*x*x

fig = plt.figure()                         
ax = fig.gca()                              

ax.set_xticks(np.arange(0, 6, 1))   
ax.set_yticks(np.arange(0, 800,50))  

plt.plot(x, y1, 'r-') 
plt.plot(x, y2, 'g--') 
plt.plot(x, y3,'b:')
plt.title('Simple Plot')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend(('y=x', '$y=x^2$','$y=x^3$'), loc='upper right')

plt.grid()
plt.show()
````

#### (2) Draw the graph of the following equations according to the given conditions.

$ x(t) = 16sin^3(t)\\
  y(t)=13cos(t)−5cos(2t)−2cos(3t)−cos(4t)$

#### Condition:
* Color: deeppink, Line: solid.
* Alpha value of line: 0.5.
* Title: Heart Curve.
* Legend: Heart Curve.
* Figure size: 6x6 inch.
* Backgroud color: light pink.


#### Tips
$t$ is an angle, so, first generate $t$ from 0 to $2\pi$ with at least 100 intervals.

````python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np

t = np.linspace(0, 2 * np.pi, 100)

x = 16 * np.sin(t)**3
y = 13 * np.cos(t) - 5 * np.cos(2 * t) - 2 * np.cos(3 * t) - np.cos(4 * t)

fig = plt.figure(figsize = (6,6))          
ax = fig.gca()
plt.plot(x, y, label='Heart Curve', color='deeppink', linestyle='-', alpha=0.5)

plt.title('Heart Curve')
plt.legend()
ax.set_facecolor('lightpink')

plt.show()
````

#### (3) Draw the graph of the following equations according to the given conditions.

$ y=x^3 $

#### Condition:
* Figure size: 10x10 inch
* Use subplot: the first one normal graph, and the second one should have the origin at the center of the graph.
* Title: "Normal Origin", and "Origin at the center"
* Legend: $y=x^3$.
* Range of X: -4 to 4 with 100 intervals
* Figure size: 8x8 inch.
* Label: X and Y (the 2nd subplot may located labels at strange places.)
* Grid on



````python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-4, 4, 100)
y = x**3

fig = plt.figure(figsize=(10, 10))

plt.subplot(211)
plt.plot(x, y, label='$y=x^3$')
plt.title('Normal Origin')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend(loc='upper right')
plt.grid()

fig2 = plt.figure(figsize=(8, 8))
ax2 = fig2.gca()
plt.plot(x, y, label='$y=x^3$')
plt.title('Origin at the center')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend(loc='upper right')
plt.grid()

ax2.spines['left'].set_position('center')
ax2.spines['bottom'].set_position('center')
ax2.spines['right'].set_color('none')
ax2.spines['top'].set_color('none')
ax2.xaxis.set_ticks_position('bottom')
ax2.yaxis.set_ticks_position('left')

plt.show()
````

#### (4) Build (Draw) the following graph in `networkx` using the functions to add graphs. 

#### Condition:
* The weight of each edge should correspond to the sum of the node labels it connects.

<img src="example-graph.png" width="300px" />


````python
%matplotlib inline
import networkx as nx
import matplotlib.pyplot as plt

G = nx.Graph()

nodes = [1, 2, 3, 4, 0]
G.add_nodes_from(nodes)

edges = [(1, 2), (2, 3), (3, 4), (4, 1)]  

for node in nodes[:-1]:
    G.add_edge(node, 0, weight=node + 0)

for u, v in edges:
    G.add_edge(u, v, weight=u + v)

pos = nx.spring_layout(G) 
plt.figure(figsize=(8, 6))
nx.draw(G, pos, with_labels=True, node_color='lightblue', node_size=3000, font_size=16, font_weight='bold')
labels = nx.get_edge_attributes(G, 'weight') 
nx.draw_networkx_edge_labels(G, pos, edge_labels=labels)

plt.title("Graph with Central Node 0 and Edge Weights as Sum of Node Labels")
plt.show()
````
